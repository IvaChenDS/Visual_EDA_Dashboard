# 1. Visualization Technique 
## 1.1 A narrative description of each visualization type in your dashboard 
### **Line Chart** : Show the "Change" over time. 
The line chart is chosen because it can clearly demonstrate the data fluctuation over time, <br> and it is easy and straightforward for humans to digest, as a higher position means a higher value and a lower position means a lower value. <br>

### **Combined Boxplot and Violin plot** : Show the distribution of data. 
The combined box and violin plot is chosen because boxplots work well together to show the distribution of the data. <br>
**Boxplot** <br>
- Provides information regarding median, quartiles, outliers, etc.
- Excellent for identifying outliers using whiskers.
- Shows indicate skewness by the position of the median and whisker lengths.<br> 

**Violin** <br>
- Displays the full probability density function, showing the shape of the data distribution, including multimodal distributions (multiple peaks).
- Helps understand whether the data is skewed or has multiple peaks, which boxplots alone may not clearly show.
- Gives a more visual representation of skewness and how data points are distributed across the range.<br>
 
### **Scatterplot matrix**: Show the relationship between variables. 
Scatterplot is a powerful visualization tool that displays pairwise relationships between multiple numerical variables in a dataset. <br>It is especially useful in exploratory data analysis. It can detect relationships between variables, identify patterns and trends, <br>and spot multicollinearity. 

### **HeatMap**: Show patterns or trends of numerical data in a large dataset conditional on the other two variables.
A heatmap visualizes numerical data in a matrix format where values are encoded using colors, making it easier to detect patterns, trends, and relationships in large datasets. <br>
It is particularly useful for exploring how one variable changes based on two other variables, effectively representing three-dimensional <br> data intuitively. This makes it easier for the human mind to digest complex relationships at a glance. <br>

## 1.2 A discussion of how these visualizations complement each other and when each should be used
### Complement each other
- Use a Line Chart for time-series trends and see the mean value for a given time point, and then further explore the data by analyzing distribution with a Boxplot & Violin Plot.  
- Scatterplot Matrices help explore relationships between variables, which can be validated with Heatmaps (e.g., correlation analysis).  
- Heatmaps summarize complex data visually, complementing scatterplots by providing a higher level view.  

### When to use
**Line Chart**:  To observe how a variable changes over time (e.g., death count change over time, stock prices, temperature changes, sales trends).  
**Boxplot**: Understand the **spread, skewness, and outliers** in a dataset and highlight quartiles, median, and outliers.  
**Violin**: See distribution shape and show density, multimodal distributions, and shape of data.  
**Scatterplot Matrix**: Explore pairwise relationships between multiple numerical variables and detect correlations, trends, multicollinearity, and clusters between variables.  <br>
**Heatmap**: Find hidden patterns and understand how one variable changes based on two other variables and visualize numerical data with color encoding, making it easier to spot trends, outliers, and relationships across a large dataset.  

## 1.3 Dashboard-specific considerations like interactivity options 
**Line Chart**: Select the variables of interest. For example, how does the death count of {Disease} change over time in a given {State}? <br> Show {the given timeframe of the data} when users move the cursor to a position on a line cart. <br>
**Box_Volin**: Select the variable of interest. For example, what is the distribution of the death count of {Disease} in a given {State} by 2020, 2021, 2023, 2024, 2025?<br>
**Scatterplot Matrix** Select relevant numerical variables and the targeted year. For example, what are the relationships between the the death count of {Disease 1}, {Disease 2}, and {Disease 3} in {Year 2020}? <br>
**Heatmap**: Choose the disease. For example, what is the trend of the death count of {Disease} from 2020 to 2025, and does the trend differ by 52 states?

## 2.1 Dash & Plotly ##

**What is Dash? What is Plotly? Are they Open source? What is Dash and Poltly used for?** <br>
Dash is an open-source framework built on top of Flask, Plotly.js, and React.js, enabling users to create interactive web applications using Python. <br>
Plotly is an open-source graphing library used for creating interactive visualizations in Python, R, JavaScript, and Julia. It is widely used for data visualization, dashboards, and web-based analytics.<br>
It supports various types of interactive charts, including:<br>
- Line charts, bar charts, scatter plots
- Heatmaps, box plots, violin plots
- 3D plots, choropleth maps, and sunburst charts
- Network graphs and financial charts <br>

**Who created Dash and Plotly?** <br>
Plotly, https://plotly.com/, a company specializing in data visualization and analytics, developed and open-sourced Dash. <br>

**How to install Dash and Polty?** <br>
Type "! pip install dash plotly" in Jupyter Notebook.<br>


**Why use this (Dash + Plotly) framework?** <br>
Dash and Plotly simplify the process of building web applications, as it abstracts away the complexities of web development, allowing data scientists<br> and analysts to focus on data analysis and visualization. Benefits incldue:<br>
- Pure Python Development – Unlike JavaScript-based frameworks, Dash allows developers to create web applications entirely in Python, without needing frontend expertise.<br>
- Seamless Integration with Plotly – Provide native support for interactive visualizations. <br>
- Full Web App Capabilities – Enable the creation of fully functional dashboard applications without needing Flask or Django for backend work.
- Interactivity & Callbacks – support highly dynamic applications through Python-defined callback functions that update components in response to user inputs.<br>
- Seamless Integration – Works well with Pandas, and NumPy.
- Automatically adds hover (tooltip) functionality.
- Cross-Language Support – Available in Python, R, and JavaScript. <br>

**Is it declarative or procedural?** <br>
Dash and Plotly primarily follow a declarative programming paradigm. Users can define the layout and UI components declaratively using Python, <br> specifying what the dashboard should look like.
The interactivity is managed through reactive callbacks, where functions automatically respond to input changes rather than requiring explicit procedural calls.

### Load and Preprocess the Data
The dataset contains death counts categorized by disease, state, and time.

````python
raw = pd.read_csv('COVID-19_Death_Counts.csv')
raw.columns
````

````python
raw['Group'].unique()
````

````python
## Only Include Week Data
df = raw[raw["Group"] == "By Week"].copy()
````

````python
# Convert "End Date" to datetime
df["End Date"] = pd.to_datetime(df["End Date"], format="%m/%d/%y")
df["Month"]= df["End Date"].dt.month
df["Year"]=df["End Date"].dt.year
df["Week"] = df["End Date"].dt.to_period("W")
### Clean Year ####
### Transform 2020/2021 io 2021; 2021/2022 to 2022
year_mapping = {
    '2019/2020': '2020',
    '2020/2021': '2021',
    '2021/2022': '2022'
}
df['Year'] = df['Year'].replace(year_mapping)
````

````python
df.columns
````

### Add Population Data to Calculate Death Rate
** Create df_death_rate_by_state**
- This df_death_rate_by_state will be used for the final interactive graph: To tell a cohesive story.
Below is a good look of population data.


````python
df_population= pd.read_csv('NST-EST2024-POP.csv')
# Clean and format the population DataFrame

df_population = df_population.rename(columns={df_population.columns[0]: "State"})

for year in range(2020, 2025):
    col = f"Population {year}"
    df_population[col] = df_population[col].astype(str).str.replace(",", "")
    df_population[col] = pd.to_numeric(df_population[col], errors="coerce")



# Remove leading dots from state names
df_population["State"] = df_population["State"].str.replace(r"^\.", "", regex=True).str.strip()

# Convert population columns to numeric
for year in range(2020, 2025):
    df_population[f"Population {year}"] = pd.to_numeric(df_population[f"Population {year}"], errors="coerce")


df_deaths_by_year = df.groupby(["State", "Year"])["Total Deaths"].sum().reset_index()


# Reshape population data to long format for merging
df_population_long = df_population.melt(
    id_vars=["State"],
    value_vars=[f"Population {year}" for year in range(2020, 2025)],
    var_name="Year",
    value_name="Population"
)
df_population_long["Year"] = df_population_long["Year"].str.extract(r"(\d{4})").astype(int)

# Merge the datasets
df_death_rate_by_state = pd.merge(
    df_deaths_by_year,
    df_population_long,
    on=["State", "Year"],
    how="inner"
)

df_death_rate_by_state["Death Rate"] = df_death_rate_by_state["Total Deaths"] / df_death_rate_by_state["Population"]

df_death_rate_by_state.head()
````

# Use Plotly Express to create line charts #

x.line(data_frame=pd.DataFrame, x=str, y=str, color=str) <br>


- Parameter	Description 
- data_frame	The Pandas DataFrame containing the data.
- x 	        The column to use for the x-axis (e.g., time or date).
- y	            The column to use for the y-axis (e.g., death rates).
- color 	    (Optional) Categorizes lines based on another column.
- title	        (Optional) Sets the chart title.

````python
# Set up input parameters Time-Series Chart
@app.callback(
    dd.Output('time-series-chart', 'figure'),
    [dd.Input('update-button', 'n_clicks')],
    [dd.State('time-disease-dropdown', 'value'),
     dd.State('time-state-dropdown', 'value')]
)
def update_time_series(n_clicks, selected_disease, selected_state): 
    if n_clicks==0:
         return dash.no_update 
    
    if selected_state not in df["State"].unique():
        return px.line(title="No Data Available")
    
    filtered_df=df[df["State"]==selected_state].copy() 
    
    # Convert 'Week' column from Period to Timestamp if needed
    if isinstance(filtered_df["Week"].iloc[0], pd.Period):
        filtered_df["Week"] = filtered_df["Week"].dt.to_timestamp()

    # Convert to datetime (ensuring it's in timestamp format)
    filtered_df["Week"] = pd.to_datetime(filtered_df["Week"])

    # Sort the DataFrame
    filtered_df = filtered_df.sort_values(by=["Week"])


 
    #if "Week" is a Period object, it cannot be plotted in Plotly because it isn't JSON serializable.
    #We need to convert it to a string or timestamp before passing it to Plotly.   

    min_y_value = filtered_df[selected_disease].min()
    max_y_value = filtered_df[selected_disease].max()
    y_min = min_y_value * 0.9 if min_y_value > 0 else 0
    y_max = max_y_value * 1.1

    time_fig = px.line(filtered_df, x="Week", y=selected_disease,
                       title=f"{selected_disease} in {selected_state} Over Time",
                       labels={"Week": "Week", selected_disease: "Deaths"},
                       hover_data={"Week": "|%Y-%m-%d"} )

    time_fig.update_yaxes(range=[y_min, y_max])

    return time_fig
````

# Use Plotly Express to create box and violin plots #

 px.violin(data_frame=pd.DataFrame, x=str, y=str, color=str, box=bool, points=str) <br>

- Argument  	Description
- data_frame	The Pandas DataFrame containing the data.
- x	            Categorical variable (e.g., "State", "Disease").
- y	            Numeric variable (e.g., "COVID-19 Deaths").
- color	        Adds color grouping (same as x).
- box	        If True, adds a boxplot inside the violin plot.
- points	    Shows individual points: "all", "outliers", "none". 



````python
# Set up input parameters for box-violin plot
@app.callback(
    dd.Output('box-violin-plot', 'figure'),
    [dd.Input('update-button2', 'n_clicks')], 
    [dd.State('box-disease-dropdown', 'value'),
    dd.State('box-state-dropdown', 'value')
    ]
)
def box_violin(n_clicks, selected_disease, selected_state):
    if n_clicks==0:
         return dash.no_update 



    df_filtered = df.dropna(subset=[selected_disease])

    filtered_df = df_filtered[(df_filtered['State'] == selected_state)].copy()

    
    if filtered_df.empty:
        return px.violin(title=f"No Data Available for {selected_disease} in {selected_state}")

        
# Boxplot & Violin Plot
    box_violin_fig = px.violin(filtered_df, x="Year", y=selected_disease, box=True, points="outliers", 
                               title=f"Violin Plot of {selected_disease} in {selected_state}")

    return box_violin_fig
````

# Use Plotly Express to create a scatterplot matrix #
px.scatter_matrix(data_frame=pd.DataFrame, dimensions=list, color=str, symbol=str, title=str)<br>

- Argument  	Description
- data_frame	The Pandas DataFrame containing the data.
- dimensions	A list of numerical columns to be compared in the scatterplot matrix.
- color	        (Optional) Column name for grouping data by color.
- symbol	    (Optional) Column name for different point markers.
- size	        (Optional) Column name to set the size of scatterplot markers.
- title	        (Optional) Title of the scatterplot matrix.



````python
# Set up Input parameters for scatter-matrix
@app.callback(
    [dd.Output('scatter-matrix', 'figure')],
    [dd.Input('update-button3', 'n_clicks')], 
    [dd.State('scatter-state-dropdown', 'value'),
    dd.State('scatter-year-dropdown', 'value')]
)


def scatter(n_clicks, selected_states, selected_years):
    if n_clicks==0:
         return dash.no_update
    
    filtered_df = df[(df['State'] == selected_states) & (df['Year'] == selected_years)].copy()
    
    if len(filtered_df) <=5:
        return px.scatter_matrix(title="No enough data available")

    diseases =['COVID-19 Deaths', 'Pneumonia Deaths', 'Influenza Deaths',
       'Total Deaths']
    
    scatter_fig = px.scatter_matrix(filtered_df, dimensions=diseases,
                            opacity=0.7,
                            labels={"value": "Deaths"},
                            title= f"Scatterplot Matrix of Deaths Count in {selected_states} in Year {selected_years}")
                            

    scatter_fig.update_layout(height=800)

    return (scatter_fig,)
````

# Create Heatmap #
Aggregate data to see sum of death per year for a given disease <br>
Use px.imshow 

px.imshow(img=np.array, labels=dict, x=list, y=list, color_continuous_scale=str, title=str)<br>

- Argument	                 Description
- img	                     Required A 2D array (list, NumPy array, or Pandas DataFrame) representing the heatmap values.
- x	                         Labels for the x-axis (e.g., column names for heatmaps).
- y	                         Labels for the y-axis (e.g., row names for heatmaps).
- labels	                 Dictionary for axis labels (labels={"color": "Deaths"}).
- color_continuous_scale	 Color scheme for heatmap ("Reds", "Viridis", "Blues", etc.).
- title	                     Sets the heatmap title.
- zmin, zmax	             Min/max values for color scale.
- origin	                 Determines the heatmap orientation ("upper" or "lower").

````python
# Set up Input parameters for heatmap
@app.callback(
     dd.Output('heatmap', 'figure'),
    [dd.Input('update-button4', 'n_clicks')], 
    [dd.State('heatmap-disease-dropdown', 'value')]
    
)

def heatmap(n_clicks, selected_disease):
    if n_clicks is None or n_clicks == 0:
        return dash.no_update
    
    filtered_df = df[df["State"] != "United States"].copy()   ## excluding the United States from the heatmap 

    # Heatmap (Aggregated)
    pivot_df = filtered_df.pivot_table(index="Year", columns="State", values=selected_disease, aggfunc="sum")
    heatmap_fig = px.imshow(pivot_df, labels=dict(color="Deaths"),
                            color_continuous_scale = 'Reds',
                            title=f"Heatmap of {selected_disease} by Year")


    heatmap_fig.update_layout(height=800)

    return heatmap_fig
````

# Create U.S. Map (Total Death)
px.choropleth(data_frame=pd.DataFrame,locations=str,locationmode=str,scope=str,color=str, <br>
    color_continuous_scale=str,hover_name=str,hover_data=dict,labels=dict,title=str) <br>


- Argument   	                Description
- data_frame		            The DataFrame containing the data to visualize
- locations		                Column with 2-letter U.S. state abbreviations
- locationmode	             	Interprets locations as U.S. state codes
- scope		                    Restricts the map to the U.S. region
- color		                    Column used to color each state based on value
- color_continuous_scale		Color scale from light to dark red based on value
- hover_name		            Column displayed as the hover label/title
- hover_data	                Extra data shown in the tooltip; with formatting for percentages
-

````python
# Set up Input parameters for heatmap
@app.callback(
    dd.Output("choropleth-map", "figure"),
    [dd.Input("update-map-button", "n_clicks")],
    [dd.State("choropleth-year-dropdown", "value")]
)


def update_choropleth(n_clicks, selected_year):
    if n_clicks == 0:
        return dash.no_update

    filtered_df = df_death_rate_by_state[df_death_rate_by_state["Year"] == selected_year]

    ## Plotly requires 2-letter postal abbreviations  
    us_state_abbrev = {
    'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR', 'California': 'CA',
    'Colorado': 'CO', 'Connecticut': 'CT', 'Delaware': 'DE', 'Florida': 'FL', 'Georgia': 'GA',
    'Hawaii': 'HI', 'Idaho': 'ID', 'Illinois': 'IL', 'Indiana': 'IN', 'Iowa': 'IA',
    'Kansas': 'KS', 'Kentucky': 'KY', 'Louisiana': 'LA', 'Maine': 'ME', 'Maryland': 'MD',
    'Massachusetts': 'MA', 'Michigan': 'MI', 'Minnesota': 'MN', 'Mississippi': 'MS', 'Missouri': 'MO',
    'Montana': 'MT', 'Nebraska': 'NE', 'Nevada': 'NV', 'New Hampshire': 'NH', 'New Jersey': 'NJ',
    'New Mexico': 'NM', 'New York': 'NY', 'North Carolina': 'NC', 'North Dakota': 'ND',
    'Ohio': 'OH', 'Oklahoma': 'OK', 'Oregon': 'OR', 'Pennsylvania': 'PA', 'Rhode Island': 'RI',
    'South Carolina': 'SC', 'South Dakota': 'SD', 'Tennessee': 'TN', 'Texas': 'TX', 'Utah': 'UT',
    'Vermont': 'VT', 'Virginia': 'VA', 'Washington': 'WA', 'West Virginia': 'WV',
    'Wisconsin': 'WI', 'Wyoming': 'WY'
}

    # Then apply mapping before plotting
    filtered_df["State Code"] = filtered_df["State"].map(us_state_abbrev)

    map = px.choropleth(
        filtered_df,
        locations="State Code",
        locationmode="USA-states",
        scope="usa",
        color="Death Rate",
        color_continuous_scale="Reds",
        hover_name="State",
        hover_data={"Death Rate": ':.2%', "Population": True, "Total Deaths": True},
        title=f"Total Death Rate by State in {selected_year}"
    )
    map.update_geos(fitbounds="locations", visible=False)
    map.update_layout(height=600)

    return map
````
